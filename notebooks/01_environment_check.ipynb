# 연구 환경 확인

주요 라이브러리와 실행 환경을 확인하고, 예시 작업자 분포의 Shannon 엔트로피를 계산합니다.

```python
import os
import platform
import sys

import matplotlib.pyplot as plt
from matplotlib import font_manager
import numpy as np
import pandas as pd
import scipy
import simpy

available_fonts = {font.name for font in font_manager.fontManager.ttflist}
korean_fonts = ["Malgun Gothic", "AppleGothic", "NanumGothic", "Noto Sans CJK KR"]
selected_font = next((font for font in korean_fonts if font in available_fonts), None)
if selected_font:
    plt.rcParams["font.family"] = selected_font
else:
    print("한글 그래프 표시를 위해 Noto Sans CJK KR 등의 한글 폰트를 설치하세요.")
plt.rcParams["axes.unicode_minus"] = False

print("주요 라이브러리를 정상적으로 불러왔습니다.")
```

```python
print("Python 버전:", platform.python_version())
print("Python 실행 경로:", sys.executable)
print("현재 작업 폴더:", os.getcwd())
print("NumPy 버전:", np.__version__)
print("pandas 버전:", pd.__version__)
print("SciPy 버전:", scipy.__version__)
print("SimPy 버전:", simpy.__version__)
```

```python
zones = ["A", "B", "C"]
worker_distribution = np.array([0.5, 0.3, 0.2])
entropy = -np.sum(worker_distribution * np.log2(worker_distribution))

print(f"작업자 분포의 Shannon 엔트로피: {entropy:.6f}")

plt.figure(figsize=(7, 4))
plt.bar(zones, worker_distribution)
plt.xlabel("피킹 구역")
plt.ylabel("작업자 비율")
plt.title("구역별 작업자 분포")
plt.ylim(0, 1)
plt.tight_layout()
plt.show()
```
