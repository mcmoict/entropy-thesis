```python
import numpy as np
import matplotlib.pyplot as plt

zones = ["A", "B", "C"]
worker_distribution = np.array([0.5, 0.3, 0.2])

entropy = -np.sum(
    worker_distribution * np.log2(worker_distribution)
)

print(f"작업자 분산 엔트로피: {entropy:.6f}")

plt.figure(figsize=(7, 4))
plt.bar(zones, worker_distribution)
plt.xlabel("Zone")
plt.ylabel("Worker Ratio")
plt.title("Worker Distribution")
plt.ylim(0, 1)
plt.tight_layout()
plt.show()
```
